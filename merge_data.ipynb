```python
import pandas as pd

# Load files
alc_df = pd.read_csv("./ALC_Export.csv", dtype=str)
geo_df = pd.read_csv("./Geography.csv", dtype=str)
occ_df = pd.read_csv("./oes_soc_occs.csv", dtype=str)

# Clean
alc_df.columns = alc_df.columns.str.strip()
geo_df.columns = geo_df.columns.str.strip()
occ_df.columns = occ_df.columns.str.strip()
occ_df = occ_df.rename(columns={"soccode": "SocCode"})

# Group geography to combine counties per Area
geo_df_grouped = (
    geo_df.groupby("Area", as_index=False)
    .agg({
        "CountyTownName": lambda x: ", ".join(sorted(set(x))),
        "AreaName": "first",
        "State": "first",
        "StateAb": "first"
    })
)

# Merge
merged_df = alc_df.merge(geo_df_grouped, on="Area", how="left")
final_df = merged_df.merge(occ_df[["SocCode", "Title"]], on="SocCode", how="left")

# Reorder
final_columns = [
    "Area", "SocCode", "GeoLvl", "Level1", "Level2", "Level3", "Level4",
    "Average", "Label", "CountyTownName", "AreaName", "State", "StateAb", "Title"
]
final_df = final_df[final_columns]

# Save to file
final_df.to_csv("ALC_with_geography_and_soc.csv", index=False)

print("✅ Merged file saved with", len(final_df), "rows.")
```
